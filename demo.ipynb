```python
from speechbrain.inference.separation import SepformerSeparation as separator
import torchaudio
```

```python
model = separator.from_hparams(source="speechbrain/resepformer-wsj02mix", savedir='pretrained_models/resepformer-wsj02mix')
```

```python
est_sources = model.separate_file(path='example.wav') 
```

```python
torchaudio.save("output1.wav", est_sources[:, :, 0].detach().cpu(), 8000)
torchaudio.save("output2.wav", est_sources[:, :, 1].detach().cpu(), 8000)
```
